### Outlier Detection Using IQR

```python
# Select numerical columns
numeric_data = df.select_dtypes(include=np.number)

Q1 = numeric_data.quantile(0.25)

Q3 = numeric_data.quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outlier_count = (
    (numeric_data < lower_limit) |
    (numeric_data > upper_limit)
).sum()

outlier_count[outlier_count > 0]
```

### Handle Numerical and Categorical Features

```python
# Create a copy of the cleaned dataset
data = df.copy()

# Remove Customer_ID because it is only an identifier
# Remove Customer_Since because we already extracted useful date features
# Remove constant columns because they do not provide information for clustering
data = data.drop(
    columns=[
        'Customer_ID',
        'Customer_Since',
        'Contact_Cost',
        'Revenue_Contribution'
    ]
)

# Display the remaining columns
data.columns
```

### Separate Numerical and Categorical Features

```python
numerical_features = data.select_dtypes(
    include=np.number
).columns

categorical_features = data.select_dtypes(
    exclude=np.number
).columns

numerical_features
```

```python
categorical_features
```

### Convert Categorical Features

```python
data_encoded = pd.get_dummies(
    data,
    columns=categorical_features,
    drop_first=True,
    dtype=int
)

# Display the encoded dataset
data_encoded.head()
```

### Standard Scaling

```python
# Import StandardScaler
from sklearn.preprocessing import StandardScaler

# Create the StandardScaler object
scaler = StandardScaler()

# Scale all numerical features
X_scaled = scaler.fit_transform(data_encoded)

# Convert the scaled array back into a DataFrame
X_scaled = pd.DataFrame(
    X_scaled,
    columns=data_encoded.columns,
    index=data_encoded.index
)

# Display the scaled features
X_scaled.head()
```


K-Means

```python
# Import K-Means clustering
from sklearn.cluster import KMeans
```

```python
# Store WCSS values for different numbers of clusters
wcss = []

# Try cluster values from 1 to 10
for k in range(1, 11):
    
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(X_scaled)
    
    # Store the WCSS value
    wcss.append(kmeans.inertia_)
```

```python
# Plot the Elbow Method
plt.figure(figsize=(9, 6))

plt.plot(
    range(1, 11),
    wcss,
    marker='o',
    linewidth=2
)

plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.title('Elbow Method for K-Means Clustering')

plt.xticks(range(1, 11))
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Set the number of clusters based on the elbow graph
k = 4

# Create the K-Means model
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

# Create cluster labels
kmeans_clusters = kmeans.fit_predict(X_scaled)

# Add cluster labels to the original dataset
df['KMeans_Cluster'] = kmeans_clusters

# Display the first 5 rows
df.head()
```

```python
# Count customers in each K-Means cluster
df['KMeans_Cluster'].value_counts().sort_index()
```

```python
# Import silhouette score
from sklearn.metrics import silhouette_score

# Calculate silhouette score
kmeans_silhouette = silhouette_score(
    X_scaled,
    kmeans_clusters
)

# Display the score
print("K-Means Silhouette Score:", round(kmeans_silhouette, 4))
print(
    "K-Means Silhouette Score (%):",
    round(kmeans_silhouette * 100, 2),
    "%"
)
```

```python
# K-Means cluster visualization

plt.figure(figsize=(10, 6))

scatter = plt.scatter(
    df['Annual_Income'],
    df['Total_Amount_Spent'],
    c=kmeans_clusters,
    cmap='viridis',
    s=30,
    alpha=0.7
)

# Add labels
plt.xlabel('Annual Income')
plt.ylabel('Total Amount Spent')
plt.title('K-Means Customer Segmentation')

# Show cluster colors
plt.colorbar(scatter, label='K-Means Cluster')

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

Hierarchical

```python
# Import Agglomerative Clustering
from sklearn.cluster import AgglomerativeClustering
```

```python
# Set the number of clusters
k = 5

# Create the Hierarchical Clustering model
hierarchical_model = AgglomerativeClustering(
    n_clusters=5,
    linkage='ward'
)

# Create cluster labels
hierarchical_clusters = hierarchical_model.fit_predict(X_scaled)

# Add cluster labels to the dataset
df['Hierarchical_Cluster'] = hierarchical_clusters

# Display the first 5 rows
df.head()
```

```python
# Count customers in each Hierarchical cluster
df['Hierarchical_Cluster'].value_counts().sort_index()
```

```python
# Import functions for creating a dendrogram
from scipy.cluster.hierarchy import dendrogram, linkage

# Create the linkage matrix using Ward's method
linkage_matrix = linkage(
    X_scaled,
    method='ward'
)

# Create the dendrogram
plt.figure(figsize=(12, 6))

dendrogram(
    linkage_matrix,
    truncate_mode='lastp',
    p=20,
    leaf_rotation=90,
    leaf_font_size=10
)

plt.xlabel('Customer Groups')
plt.ylabel('Distance')
plt.title('Hierarchical Clustering Dendrogram')

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Import silhouette score
from sklearn.metrics import silhouette_score

# Calculate silhouette score
hierarchical_silhouette = silhouette_score(
    X_scaled,
    hierarchical_clusters
)

# Display the score
print(
    "Hierarchical Silhouette Score:",
    round(hierarchical_silhouette, 4)
)

print(
    "Hierarchical Silhouette Score (%):",
    round(hierarchical_silhouette * 100, 2),
    "%"
)
```

```python
# Hierarchical clustering visualization

plt.figure(figsize=(10, 6))

scatter = plt.scatter(
    df['Annual_Income'],
    df['Total_Amount_Spent'],
    c=hierarchical_clusters,
    cmap='plasma',
    s=50,
    alpha=0.7
)

# Add labels
plt.xlabel('Annual Income')
plt.ylabel('Total Amount Spent')
plt.title('Hierarchical Clustering - Customer Segmentation')

# Show cluster colors
plt.colorbar(
    scatter,
    label='Hierarchical Cluster'
)

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

DBSCAN

```python
# Import DBSCAN
from sklearn.cluster import DBSCAN
```

```python
# Create the DBSCAN model
dbscan_model = DBSCAN(
    eps=0.5,
    min_samples=5
)

# Create cluster labels
dbscan_clusters = dbscan_model.fit_predict(X_scaled)

# Add DBSCAN cluster labels to the dataset
df['DBSCAN_Cluster'] = dbscan_clusters

# Display the first 5 rows
df.head()
```

```python
# Count customers in each DBSCAN cluster
df['DBSCAN_Cluster'].value_counts().sort_index()
```

```python
# Import NearestNeighbors
from sklearn.neighbors import NearestNeighbors

# Set the number of nearest neighbors
k = 5

# Create NearestNeighbors model
neighbors = NearestNeighbors(n_neighbors=k)

# Fit the model
neighbors_fit = neighbors.fit(X_scaled)

# Calculate distances
distances, indices = neighbors_fit.kneighbors(X_scaled)

# Get the distance to the 5th nearest neighbor
k_distances = distances[:, k - 1]

# Sort the distances
k_distances = np.sort(k_distances)

# Plot the K-Distance Graph
plt.figure(figsize=(10, 6))

plt.plot(
    k_distances,
    linewidth=2
)

plt.xlabel('Customers Sorted by Distance')
plt.ylabel('5th Nearest Neighbor Distance')
plt.title('K-Distance Graph for DBSCAN')

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# DBSCAN cluster visualization

plt.figure(figsize=(10, 6))

scatter = plt.scatter(
    df['Annual_Income'],
    df['Total_Amount_Spent'],
    c=dbscan_clusters,
    cmap='plasma',
    s=50,
    alpha=0.7
)

plt.xlabel('Annual Income')
plt.ylabel('Total Amount Spent')
plt.title('DBSCAN Customer Segmentation')

plt.colorbar(
    scatter,
    label='DBSCAN Cluster'
)

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

PCA

```python
# Import PCA
from sklearn.decomposition import PCA
```

```python
# Create PCA with 2 components
pca = PCA(n_components=2)

# Transform the scaled data into 2 principal components
X_pca = pca.fit_transform(X_scaled)

# Check the new shape
X_pca.shape
```

```python
# Explained variance of each principal component
print(
    "PC1 Explained Variance:",
    round(pca.explained_variance_ratio_[0] * 100, 2),
    "%"
)

print(
    "PC2 Explained Variance:",
    round(pca.explained_variance_ratio_[1] * 100, 2),
    "%"
)

# Total variance explained
print(
    "Total Explained Variance:",
    round(pca.explained_variance_ratio_.sum() * 100, 2),
    "%"
)
```

```python
# PCA 2D visualization

plt.figure(figsize=(10, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    cmap='viridis',
    s=50,
    alpha=0.7
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA - 2D Visualization')

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# PCA visualization with K-Means clusters

plt.figure(figsize=(10, 6))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=kmeans_clusters,
    cmap='viridis',
    s=50,
    alpha=0.7
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA Visualization of K-Means Customer Clusters')

plt.colorbar(
    scatter,
    label='K-Means Cluster'
)

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Create PCA with 3 components
pca_3d = PCA(n_components=3)

# Transform the scaled data into 3 principal components
X_pca_3d = pca_3d.fit_transform(X_scaled)

# Check the new shape
X_pca_3d.shape
```
